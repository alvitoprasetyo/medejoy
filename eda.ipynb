```python
kolom_num = df.select_dtypes(include=['float64', 'int64']).columns
```

```python
# Histogram + boxplot per kolom
for col in kolom_num:
    fig, axes = plt.subplots(1, 2, figsize=(14, 4))

    # Histogram
    axes[0].hist(df[col].dropna(), bins=50, color='steelblue', edgecolor='white')
    axes[0].set_title(f'Histogram: {col}')
    axes[0].set_xlabel(col)
    axes[0].set_ylabel('Frekuensi')

    # Boxplot
    axes[1].boxplot(df[col].dropna())
    axes[1].set_title(f'Boxplot: {col}')
    axes[1].set_ylabel(col)

    plt.tight_layout()
    plt.show()
```

```python
print(df['IS_PROMOTION'].value_counts())
print(f"\n% Promo: {df['IS_PROMOTION'].mean()*100:.2f}%")

# Pie chart
df['IS_PROMOTION'].value_counts().plot(
    kind='pie', autopct='%1.2f%%',
    colors=['#66b3ff', '#ff9999'],
    labels=['Non-Promo', 'Promo']
)
plt.title('Proporsi Promo vs Non-Promo')
plt.ylabel('')
plt.show()
```

```python
df['DATE'] = pd.to_datetime(df['DATE'])

# Rentang
print("Min DATE:", df['DATE'].min())
print("Max DATE:", df['DATE'].max())
print("Jumlah hari unik:", df['DATE'].dt.date.nunique())

# Distribusi per tahun
df['DATE'].dt.year.value_counts().sort_index().plot(kind='bar', color='steelblue')
plt.title('Distribusi Transaksi per Tahun')
plt.xlabel('Tahun')
plt.ylabel('Jumlah Transaksi')
plt.tight_layout()
plt.show()

# Distribusi per bulan
df['DATE'].dt.month.value_counts().sort_index().plot(kind='bar', color='steelblue')
plt.title('Distribusi Transaksi per Bulan (agregat semua tahun)')
plt.xlabel('Bulan')
plt.ylabel('Jumlah Transaksi')
plt.tight_layout()
plt.show()

# Distribusi per hari
urutan_hari = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
df['DATE'].dt.day_name().value_counts().reindex(urutan_hari).plot(kind='bar', color='steelblue')
plt.title('Distribusi Transaksi per Hari')
plt.xlabel('Hari')
plt.ylabel('Jumlah Transaksi')
plt.tight_layout()
plt.show()
```

```python
# Heatmap tahun x bulan
pivot = df.groupby([df['DATE'].dt.year, df['DATE'].dt.month]).size().unstack()

plt.figure(figsize=(12, 6))
sns.heatmap(pivot, cmap='YlOrRd', annot=True, fmt='.0f', cbar_kws={'label': 'Jumlah Transaksi'})
plt.title('Heatmap Transaksi: Tahun × Bulan')
plt.xlabel('Bulan')
plt.ylabel('Tahun')
plt.tight_layout()
plt.show()
```

```python
df['PERIODE'] = df['DATE'].dt.to_period('M').astype(str)

pivot = df.groupby(['DISTRIBUTOR', 'PERIODE']).size().unstack(fill_value=0)

# Top 20 distributor
top_dist = df['DISTRIBUTOR'].value_counts().head(20).index
pivot_top = pivot.loc[top_dist]

plt.figure(figsize=(20, 8))
sns.heatmap(pivot_top > 0, cmap='RdYlGn', cbar=False,
            linewidths=0.3, linecolor='gray')
plt.title('Coverage Distributor × Bulan')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df['PERIODE'] = df['DATE'].dt.to_period('M').astype(str)

trend = df.groupby('PERIODE')['GROSS_SALES'].sum().reset_index()
trend['PERIODE'] = pd.to_datetime(trend['PERIODE'])

plt.figure(figsize=(15, 5))
plt.plot(trend['PERIODE'], trend['GROSS_SALES'], marker='o', color='steelblue', linewidth=2)
plt.title('Trend Total Gross Sales per Bulan')
plt.xlabel('Periode')
plt.ylabel('Gross Sales (Rp)')
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'{x/1e9:.1f}M'))
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# 1. Filter data dari tahun 2023 ke atas
df_filtered = df[df['YEAR'] >= 2023].copy()

# 2. Buat kolom periode Bulan-Tahun (YYYY-MM) dari kolom DATE
df_filtered['YEAR_MONTH'] = df_filtered['DATE'].dt.to_period('M')

# 3. Agregasi total GROSS_SALES dan QTY_CARTON berdasarkan Bulan dan CHANNEL
summary_df = (
    df_filtered.groupby(['YEAR_MONTH', 'CHANNEL'])[['GROSS_SALES', 'QTY_CARTON']]
    .sum()
    .reset_index()
)

# Ubah format YEAR_MONTH ke string
summary_df['YEAR_MONTH'] = summary_df['YEAR_MONTH'].astype(str)

# Rename kolom agar lebih rapi
summary_df.rename(
    columns={
        'YEAR_MONTH': 'PERIODE_BULAN',
        'CHANNEL': 'CHANNEL',
        'GROSS_SALES': 'TOTAL_GROSS_SALES',
        'QTY_CARTON': 'TOTAL_QTY_CARTON',
    },
    inplace=True,
)

# 4. Buat tabel pivot per CHANNEL
pivot_sales = summary_df.pivot(
    index='PERIODE_BULAN', columns='CHANNEL', values='TOTAL_GROSS_SALES'
).fillna(0)
pivot_qty = summary_df.pivot(
    index='PERIODE_BULAN', columns='CHANNEL', values='TOTAL_QTY_CARTON'
).fillna(0)

# 5. Simpan ke CSV (3 file terpisah agar tidak butuh openpyxl)
summary_df.to_csv('Summary_Long_Format_2023_2026.csv', index=False)
pivot_sales.to_csv('Pivot_Gross_Sales_2023_2026.csv')
pivot_qty.to_csv('Pivot_Qty_Carton_2023_2026.csv')

print('File CSV berhasil dibuat dan disimpan!')
```
